```python
import glob
import os
import numpy as np

from inflammation import models, views
```

```python
class CSVDataSource():
    def __init__(self, data_dir):
        self.data_dir = data_dir

    def load_data(self):#, data_dir):
        data_file_paths = glob.glob(os.path.join(self.data_dir, 'inflammation*.csv'))
        if len(data_file_paths) == 0:
            raise ValueError(f"No inflammation data CSV files found in path {data_dir}")
        return map(models.load_csv, data_file_paths)

class JSONDataSource():
    def __init__(self, data_dir):
        self.data_dir = data_dir
        
    def load_data(self):
        data_file_paths = glob.glob(os.path.join(self.data_dir, 'inflammation*.json'))
        if len(data_file_paths) == 0:
            raise ValueError(f"No inflammation data JSON files found in path {data_dir}")
        return map(models.load_json, data_file_paths)





```

```python
def gen_standard_deviation(data):
    means_by_day = map(models.daily_mean, data)
    means_by_day_matrix = np.stack(list(means_by_day))

    daily_standard_deviation = np.std(means_by_day_matrix, axis=0)
    return daily_standard_deviation

    
def analyse_data(data_source):
    """Calculates the standard deviation by day between datasets.

    Gets all the inflammation data from CSV files within a directory,
    works out the mean inflammation value for each day across all datasets,
    then plots the graphs of standard deviation of these means."""
    data = data_source.load_data()


    return gen_standard_deviation(data)

    
def plot_data(data, name='standard deviation by day'):
    graph_data = {
        name: gen_standard_deviation(data),
    }
    views.visualize(graph_data)
```

```python

data_source = CSVDataSource(data_dir="data")
analyse_data(data_source)
```
